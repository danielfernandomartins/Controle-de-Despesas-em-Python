```python
despesas = []


def cadastrar_despesa():
    descricao = input("Descrição da despesa: ")
    categoria = input("Categoria: ")
    valor = float(input("Valor da despesa: R$ "))

    despesa = {
        "descricao": descricao,
        "categoria": categoria,
        "valor": valor
    }

    despesas.append(despesa)

    print("\nDespesa cadastrada com sucesso!")


def listar_despesas():
    if len(despesas) == 0:
        print("\nNenhuma despesa cadastrada.")
        return

    print("\n=== DESPESAS CADASTRADAS ===")

    for indice, despesa in enumerate(despesas, start=1):
        print(
            f"{indice} - "
            f"{despesa['descricao']} | "
            f"Categoria: {despesa['categoria']} | "
            f"Valor: R$ {despesa['valor']:.2f}"
        )


def calcular_total():
    total = 0

    for despesa in despesas:
        total = total + despesa["valor"]

    print(f"\nTotal de despesas: R$ {total:.2f}")


def buscar_por_categoria():
    categoria_busca = input("Digite a categoria que deseja buscar: ")

    encontrou = False

    print(f"\n=== DESPESAS DA CATEGORIA: {categoria_busca} ===")

    for despesa in despesas:
        if despesa["categoria"].lower() == categoria_busca.lower():
            print(
                f"{despesa['descricao']} - "
                f"R$ {despesa['valor']:.2f}"
            )

            encontrou = True

    if encontrou == False:
        print("Nenhuma despesa encontrada nesta categoria.")


while True:

    print("\n==============================")
    print("   CONTROLE DE DESPESAS")
    print("==============================")

    print("1 - Cadastrar despesa")
    print("2 - Listar despesas")
    print("3 - Mostrar total de despesas")
    print("4 - Buscar por categoria")
    print("5 - Sair")

    opcao = input("\nEscolha uma opção: ")

    if opcao == "1":
        cadastrar_despesa()

    elif opcao == "2":
        listar_despesas()

    elif opcao == "3":
        calcular_total()

    elif opcao == "4":
        buscar_por_categoria()

    elif opcao == "5":
        print("\nPrograma encerrado.")
        break

    else:
        print("\nOpção inválida.")
```
